```python
import pandas as pd
```

```python
train_df = pd.read_csv('/content/sample_data/california_housing_train.csv')
test_df = pd.read_csv('/content/sample_data/california_housing_test.csv')
```

```python
train_df.shape
```

```python
test_df.shape
```

```python
train_df.head(1)
```

```python
y_train = train_df['median_house_value']
X_train = train_df.drop(['median_house_value'], axis=1)

y_test = test_df['median_house_value']
X_test = test_df.drop(['median_house_value'], axis=1)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from sklearn.preprocessing import StandardScaler
```

```python
std_scaler = StandardScaler()

std_scaler.fit(X_train)

X_train_std = std_scaler.transform(X_train)
X_test_std = std_scaler.transform(X_test)
```

```python
dnn_regression_model = Sequential()
dnn_regression_model.add(Input(shape=(X_train.shape[1],)))
dnn_regression_model.add(Dense(5, activation='relu'))
dnn_regression_model.add(Dense(10, activation='relu'))
dnn_regression_model.add(Dense(1, activation='relu'))
```

```python
dnn_regression_model.compile(optimizer='adam',
              loss='mse',
              metrics=['r2_score'])
```

```python
history = dnn_regression_model.fit(X_train_std,
                   y_train,
                   epochs=100,
                   validation_data=(X_test_std, y_test))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(4, 4))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Loss Curves')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

```python
dnn_regression_model.save('house_price_model.keras')
```
