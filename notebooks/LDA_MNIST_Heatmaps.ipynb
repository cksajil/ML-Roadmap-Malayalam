# LDA on MNIST Dataset
This notebook demonstrates how to apply Linear Discriminant Analysis (LDA) to the MNIST dataset and visualize the digit class distributions as heatmaps.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.preprocessing import StandardScaler

# Load MNIST dataset
mnist = fetch_openml('mnist_784', version=1, as_frame=False)
X, y = mnist["data"], mnist["target"].astype(int)

# Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply LDA with 9 components (we have 10 digit classes)
lda = LinearDiscriminantAnalysis(n_components=9)
X_lda = lda.fit_transform(X_scaled, y)

# Calculate the mean LDA vector for each digit
lda_means = np.array([X_lda[y == i].mean(axis=0) for i in range(10)])

# Plot the mean LDA vectors as 3x3 heatmaps
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
for i, ax in enumerate(axes.flat):
    heatmap = np.zeros((3, 3))
    heatmap[:lda_means.shape[1]//3, :3] = lda_means[i][:9].reshape(-1, 3)
    ax.imshow(heatmap, cmap='viridis', interpolation='nearest')
    ax.set_title(f'Digit {i}')
    ax.axis('off')

plt.suptitle("LDA Mean Projections of MNIST Digits (as 3x3 heatmaps)", fontsize=14)
plt.tight_layout()
plt.show()
```
